## 3. Verify the file list and inspect the CSV headers

This cell checks that the eight files have compatible column names. The original files contain leading spaces in some column names, so we strip surrounding whitespace from the column labels after reading.


```python
if len(csv_files) != 8:
    raise ValueError(f"Expected 8 CSV files, but found {len(csv_files)}.")

header_info = []
reference_columns = None

for file in csv_files:
    header = pd.read_csv(file, nrows=0, encoding="latin1").columns
    cleaned_header = header.astype(str).str.strip().tolist()
    header_info.append({
        "file": file.name,
        "number_of_columns": len(cleaned_header),
        "has_label_column": "Label" in cleaned_header
    })
    if reference_columns is None:
        reference_columns = cleaned_header
    elif cleaned_header != reference_columns:
        print("Column mismatch detected in:", file.name)

header_df = pd.DataFrame(header_info)
display(header_df)

print("Number of columns in first file:", len(reference_columns))
print("Label column present:", "Label" in reference_columns)
print("First 12 column names:", reference_columns[:12])
print("Last 12 column names:", reference_columns[-12:])
```

## 7. Scan all eight files in chunks for row counts, missing values, and labels

This cell reads every row, but processes only 100,000 rows at a time to reduce memory use. It records total rows, missing values, and label counts for each file. It does not combine the entire 1.2 GB dataset into one DataFrame.


```python
CHUNK_SIZE = 100_000

file_summaries = []
missing_totals = {}
label_totals = {}

for file in csv_files:
    row_count = 0
    file_missing = {}
    file_labels = {}

    for chunk in pd.read_csv(
        file,
        chunksize=CHUNK_SIZE,
        encoding="latin1",
        low_memory=False
    ):
        chunk.columns = chunk.columns.astype(str).str.strip()
        row_count += len(chunk)

        missing_in_chunk = chunk.isna().sum()
        for column, count in missing_in_chunk.items():
            file_missing[column] = file_missing.get(column, 0) + int(count)

        if "Label" in chunk.columns:
            labels = chunk["Label"].astype("string").str.strip().value_counts(dropna=False)
            for label, count in labels.items():
                file_labels[str(label)] = file_labels.get(str(label), 0) + int(count)

    file_summaries.append({
        "file": file.name,
        "rows": row_count,
        "columns": len(reference_columns),
        "file_size_MB": round(file.stat().st_size / (1024**2), 2)
    })

    for column, count in file_missing.items():
        missing_totals[column] = missing_totals.get(column, 0) + count

    for label, count in file_labels.items():
        label_totals[label] = label_totals.get(label, 0) + count

file_summary_df = pd.DataFrame(file_summaries)
display(file_summary_df)

print("Total records across all files:", f"{file_summary_df['rows'].sum():,}")
print("Total columns (expected):", len(reference_columns))
```

```python
# Select numerical columns
numeric_df = web_df_final.select_dtypes(include=['number'])

# Generate descriptive statistics
descriptive_stats = numeric_df.describe().T

# Display results
display(descriptive_stats)
```

```python
import numpy as np
import pandas as pd

# 1. Count infinite values
infinite_counts = np.isinf(numeric_df).sum()
infinite_counts = infinite_counts[infinite_counts > 0]

print("Columns containing infinite values:")
display(infinite_counts)

# 2. Count negative values
negative_counts = (numeric_df < 0).sum()
negative_counts = negative_counts[negative_counts > 0]

print("\nColumns containing negative values:")
display(negative_counts.sort_values(ascending=False))

# 3. Identify columns with no variation
constant_columns = [
    col for col in numeric_df.columns
    if numeric_df[col].nunique(dropna=True) <= 1
]

print("\nColumns with one unique value or fewer:")
print(constant_columns)
print("\nNumber of constant columns:", len(constant_columns))
```

```python
import numpy as np

# Replace positive and negative infinity with missing values
web_df_final.replace([np.inf, -np.inf], np.nan, inplace=True)

# Check missing values after replacement
print("Missing values after replacing infinity:")

missing_values = web_df_final.isnull().sum()
display(missing_values[missing_values > 0])

# Check the actual negative values in selected columns
columns_to_check = [
    'Flow Duration',
    'Flow IAT Min',
    'Flow IAT Mean',
    'Flow IAT Max',
    'Flow Bytes/s',
    'Flow Packets/s',
    'Init_Win_bytes_forward',
    'Init_Win_bytes_backward'
]

for column in columns_to_check:
    print(f"\n{column}")
    print(web_df_final.loc[web_df_final[column] < 0, column]
          .value_counts().head(10))
```

```python
# Features where negative measurements are invalid
invalid_negative_columns = [
    'Flow Duration',
    'Flow IAT Min',
    'Flow IAT Mean',
    'Flow IAT Max',
    'Flow Bytes/s',
    'Flow Packets/s'
]

# Count negative values before cleaning
print("Negative values before cleaning:")

for column in invalid_negative_columns:
    print(column, (web_df_final[column] < 0).sum())

# Replace negative measurements with missing values
for column in invalid_negative_columns:
    web_df_final.loc[web_df_final[column] < 0, column] = np.nan

# Verify the cleaning
print("\nNegative values after cleaning:")

for column in invalid_negative_columns:
    print(column, (web_df_final[column] < 0).sum())
```
